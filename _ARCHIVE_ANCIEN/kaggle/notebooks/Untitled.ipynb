```python
import nltk
import re
from nltk.corpus import stopwords
nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

def count_word_frequency_no_stopwords(text):
    words = re.findall(r'\b\w+\b', text.lower())
    filtered = [w for w in words if w not in stop_words]
    return Counter(filtered)
```

```python
import pandas as pd
TRAIN_PATH = 'INPUT/train.csv'
TEST_PATH  = 'INPUT/test.csv'
df_train = pd.read_csv(TRAIN_PATH)
df_test  = pd.read_csv(TEST_PATH)
```

```python

text = df_train["body"]
#text = df_test["negative_example_1"]
text=text.tolist()
count_word_frequency_no_stopwords(" ".join(text))
```

```python
def extract_urls(text):
    # Pattern matches http:// or https:// followed by non-space characters
    pattern = r'https?://[^\s<>"]+'
    return re.findall(pattern, text)
```

```python
def clean_words(text):
    text = re.sub(r'^br$', ' ', text)
    text = text.lower()
    text = re.sub(r'^[a-z]\s+', ' ', text)
    text = re.sub(r'\s+', ' ', text)
    text = text.strip()   
    return text
```

```python
import re

def extract_emojis(text):
    emoji_pattern = re.compile(
        "["

        u"\U0001F600-\U0001F64F"  # Emoticons
        u"\U0001F300-\U0001F5FF"  # Symbols & pictographs
        u"\U0001F680-\U0001F6FF"  # Transport & map symbols
        u"\U0001F1E0-\U0001F1FF"  # Flags
        u"\U00002700-\U000027BF"  # Dingbats
        u"\U0001F900-\U0001F9FF"  # Supplemental symbols and pictographs
        u"\U0001FA70-\U0001FAFF"  # Extended pictographs

        "]+", flags=re.UNICODE
    )
    return emoji_pattern.findall(text)
```

```python
emojis = extract_emojis(single_line)
print(emojis)
```
