```python
import pandas as pd
TRAIN_PATH = 'INPUT/train.csv'
TEST_PATH  = 'INPUT/test.csv'
df_train = pd.read_csv(TRAIN_PATH)
df_test  = pd.read_csv(TEST_PATH)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    df_train.body, df_train.rule_violation, test_size=0.2
)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

v = CountVectorizer()  # ← parentheses needed to create an instance
X_train_cv = v.fit_transform(X_train.values.astype(str))  # ensure strings


X_test_cv = v.fit_transform(X_test.values.astype(str))  # ensure strings
```

```python
from sklearn.naive_bayes import MultinomialNB

model = MultinomialNB()
model.fit(X_train_cv, y_train)
```

```python
from sklearn.metrics import classification_report
y_pred = model.predict(X_test_cv)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report

# 1. Split your data
X_train, X_test, y_train, y_test = train_test_split(df_train.body, df_train.rule_violation, test_size=0.2, random_state=42)

# 2. Vectorize text data
vectorizer = CountVectorizer()
X_train_cv = vectorizer.fit_transform(X_train)
X_test_cv = vectorizer.transform(X_test)

# 3. Train model
model = MultinomialNB()
model.fit(X_train_cv, y_train)

# 4. Predict
y_pred = model.predict(X_test_cv)

# 5. Evaluate
print(classification_report(y_test, y_pred))
```
