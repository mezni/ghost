```python
import pandas as pd
TRAIN_PATH = 'INPUT/train.csv'
TEST_PATH  = 'INPUT/test.csv'
df_train = pd.read_csv(TRAIN_PATH)
df_test  = pd.read_csv(TEST_PATH)
```

```python
import spacy
nlp = spacy.load("en_core_web_lg")
```

```python
df = df_train
df["vector"] = df["body"].apply(lambda x: nlp(x).vector)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df.vector.values, df_train.rule_violation, test_size=0.2, random_state=2025)
```

```python
import numpy as np

#X_train.shape
X_train_2d = np.stack(X_train)
X_test_2d = np.stack(X_test)
#X_train_2d.shape
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
X_train_2d_scale = scaler.fit_transform(X_train_2d)
X_test_2d_scale = scaler.transform(X_test_2d)


model = MultinomialNB()
model.fit(X_train_2d_scale, y_train)

# 4. Predict
y_pred = model.predict(X_test_2d_scale)

# 5. Evaluate
print(classification_report(y_test, y_pred))
```
