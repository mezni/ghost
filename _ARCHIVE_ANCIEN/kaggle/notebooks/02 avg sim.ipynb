```python
import pandas as pd
TRAIN_PATH = 'INPUT/train.csv'
TEST_PATH  = 'INPUT/test.csv'
df_train = pd.read_csv(TRAIN_PATH)
df_test  = pd.read_csv(TEST_PATH)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def compute_similarity(a, b):
    vectorizer = TfidfVectorizer()
    tfidf = vectorizer.fit_transform([a, b])
    return cosine_similarity(tfidf[0:1], tfidf[1:2])[0][0]
```

```python
df_test['sim1'] = df_test.apply(lambda row: compute_similarity(row['body'], row['positive_example_1']), axis=1)
df_test['sim2'] = df_test.apply(lambda row: compute_similarity(row['body'], row['positive_example_2']), axis=1)
df_test['sim3'] = df_test.apply(lambda row: compute_similarity(row['body'], row['negative_example_1']), axis=1)
df_test['sim4'] = df_test.apply(lambda row: compute_similarity(row['body'], row['negative_example_2']), axis=1)
```

```python
submission = pd.DataFrame({
    "row_id": df_test["row_id"],
    "rule_violation": df_test[['sim1', 'sim2', 'sim3', 'sim4']].mean(axis=1)
})
```

```python
submission.to_csv("submission.csv", index=False)
submission
```
