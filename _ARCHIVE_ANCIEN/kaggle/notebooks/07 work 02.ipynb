```python
import re
import pandas as pd
```

```python
TRAIN_PATH = 'INPUT/train.csv'
TEST_PATH  = 'INPUT/test.csv'
df_train = pd.read_csv(TRAIN_PATH)
df_test  = pd.read_csv(TEST_PATH)
```

```python
def process_urls(text):
    http_pattern = r"http://[^\s]+"
    https_pattern = r"https://[^\s]+"
    
    http_urls = re.findall(http_pattern, text)
    https_urls = re.findall(https_pattern, text)

    urls = {
        'url_http_count': len(http_urls),
        'url_https_count': len(https_urls),
        'url_alpha_count': sum(len(item) for item in http_urls + https_urls),
        'url_alpha_avg': sum(len(item) for item in http_urls + https_urls)/len(https_urls)+1,
        'url_digits': sum(char.isdigit() for item in http_urls + https_urls for char in item),
        'url_params': sum(item.count("&") for item in http_urls + https_urls),
    #    'url_special_char': sum(1 for item in http_urls + https_urls for char in item if not char.isdigit() and not char.isalpha()),
    }
    return urls     

def get_features(text):
    text = text.replace('\n', ' ')
    text = re.sub(r'\s+', ' ', text).strip()
    urls = process_urls(text)

    print (urls)
    features = text
    return features
cell_str = str(df_train.loc[4, 'body'])
print (cell_str)
print('------')
features = get_features(cell_str)
print (features)
```
